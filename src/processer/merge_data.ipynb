```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import os 

os.chdir('../..')

BASE_DIR = os.getcwd()

properties_path = BASE_DIR + '/data/process/merge_properties_data.csv'
amenities_path = BASE_DIR + '/data/process/merged_amenities_data.csv'
```

```python
df_properties = pd.read_csv(properties_path, index_col=False)

df_properties.head()
```

```python
df_amenities = pd.read_csv(amenities_path, index_col=False)

df_amenities.head()
```

```python

df_final = df_properties.merge(df_amenities, left_on='id', right_on='Ma_BDS', how='left')

df_final.to_csv(BASE_DIR+'/data/process/final_dataset_for_model.csv', index=False, encoding='utf-8-sig')
```
