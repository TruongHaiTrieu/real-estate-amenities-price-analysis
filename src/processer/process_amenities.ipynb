```python
import pandas as pd
import os
import random

RAW_AMENITIES_DIR = '../../data/raw/amenities/'
PROCESS_AMENITIES_DIR = '../../data/process/amenities/'

# Từ điển tọa độ để map siêu tốc
hcm_districts = {
    'quận 1': (10.7756, 106.7019), 'q1': (10.7756, 106.7019), 'q.1': (10.7756, 106.7019),
    'quận 2': (10.7872, 106.7495), 'quận 3': (10.7843, 106.6839), 'quận 4': (10.7578, 106.7012),
    'quận 5': (10.7540, 106.6633), 'quận 6': (10.7480, 106.6352), 'quận 7': (10.7324, 106.7301),
    'quận 8': (10.7224, 106.6279), 'quận 9': (10.8428, 106.8286), 'quận 10': (10.7743, 106.6669),
    'quận 11': (10.7628, 106.6444), 'quận 12': (10.8671, 106.6413), 'bình thạnh': (10.8105, 106.7091), 
    'gò vấp': (10.8271, 106.6764), 'phú nhuận': (10.7991, 106.6788), 'tân bình': (10.8014, 106.6525), 
    'tân phú': (10.7904, 106.6254), 'bình tân': (10.7652, 106.6038), 'thủ đức': (10.8494, 106.7537), 
    'hóc môn': (10.8841, 106.5940), 'củ chi': (11.0066, 106.5022), 'nhà bè': (10.6952, 106.7317), 
    'bình chánh': (10.6873, 106.5946), 'cần giờ': (10.4079, 106.9069)
}

def add_lat_lon_to_csv(file_name):
    print(f"Đang xử lý tiện ích: {file_name}...")
    input_path = os.path.join(RAW_AMENITIES_DIR, file_name)
    df = pd.read_csv(input_path)
    
    df.columns = [c.strip().lower() for c in df.columns]
    
    def get_smart_coords(addr):
        addr_lower = str(addr).lower()
        for key, coords in hcm_districts.items():
            if key in addr_lower:
                return pd.Series([coords[0] + random.uniform(-0.01, 0.01), coords[1] + random.uniform(-0.01, 0.01)])
        return pd.Series([None, None])

    address_col = next((col for col in df.columns if 'địa chỉ' in col or 'dia_chi' in col or 'address' in col), None)
    
    if address_col:
        df[['latitude', 'longitude']] = df[address_col].apply(get_smart_coords)
    else:
        df['latitude'] = None
        df['longitude'] = None
        
    output_path = os.path.join(PROCESS_AMENITIES_DIR, file_name)
    df.to_csv(output_path, index=False)
    print(f"✅ Đã lưu tại: {output_path}\n")

amenities_files = [f for f in os.listdir(RAW_AMENITIES_DIR) if f.endswith('.csv')]
for file in amenities_files:
    add_lat_lon_to_csv(file)
```
